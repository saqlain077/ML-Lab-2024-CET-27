```python

def caught_speeding(speed, is_birthday):
    if is_birthday:
        speeding = speed - 5
    else:
        speeding = speed

    if speeding > 80:
        return 'Big Ticket'
    elif speeding > 60:
        return 'Small Ticket'
    else:
        return 'No Ticket'
print (caught_speeding(81, False))
print (caught_speeding(81, True))
print (caught_speeding(61, False))
print (caught_speeding(61, True))
print (caught_speeding(40, False))
print (caught_speeding(40, True))
```
