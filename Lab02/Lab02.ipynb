```python
import numpy as np

# REPLACE with your lab data (population in millions)
units = np.array(["Punjab", "Sindh", "KP", "Balochistan", "ICT", "AJK"])
pop23 = np.array([127.7, 55.7, 40.9, 14.9, 2.4, 4.1])
pop17 = np.array([110.0, 47.9, 35.5, 12.3, 2.0, 4.0])

print("T6a: Total =", pop23.sum(), "million")

share = pop23 / pop23.sum() * 100
print("T6b: Share (%) =", np.round(share, 2))

print("T6c: Most populous =", units[np.argmax(pop23)])
print("T6d: Units above 20 million =", units[pop23 > 20])

growth = ((pop23 / pop17) ** (1 / 6) - 1) * 100
print("T6e: Annual growth 2017-2023 (%) =", np.round(growth, 2))

print("T6f: Fastest growing =", units[np.argmax(growth)])

ajk = np.where(units == "AJK")[0][0]
print("T6g: AJK share (%) =", round(share[ajk], 2), "| AJK growth (%) =", round(growth[ajk], 2))
```

```python
# REPLACE: temp and rain are (5 cities x 12 months), Jan to Dec
cities = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Muzaffarabad"])
temp = np.array([
    [14, 17, 23, 29, 34, 36, 33, 32, 31, 27, 20, 15],
    [19, 21, 26, 30, 32, 32, 30, 29, 30, 30, 26, 21],
    [11, 13, 18, 24, 30, 34, 31, 29, 28, 23, 17, 12],
    [ 5,  8, 13, 18, 23, 27, 29, 28, 24, 17, 10,  6],
    [ 9, 11, 16, 22, 27, 31, 30, 29, 27, 22, 15, 10]])
rain = np.array([
    [20, 30, 40, 25, 20, 50, 200, 180, 80, 10, 5, 12],
    [ 8, 10, 10,  5,  2, 10,  80,  50, 20,  2, 1,  5],
    [60, 70, 90, 50, 30, 80, 300, 280, 120, 20, 10, 40],
    [30, 35, 40, 20, 8,  5,  25,  20,  5,  3,  8, 20],
    [90, 100, 130, 90, 60, 100, 350, 320, 150, 40, 25, 60]])

print("T7a: Shape =", temp.shape)
print("T7b: Yearly mean per city (°C) =", np.round(temp.mean(axis=1), 1))
print("T7c: Hottest month per city (1=Jan) =", temp.argmax(axis=1) + 1)

monthly_avg = temp.mean(axis=0)
print("T7d: 5-city monthly average (°C) =", np.round(monthly_avg, 1))
print("T7e: Months above 35 °C =", np.sum(monthly_avg > 35))

lahore_f = temp[0] * 9 / 5 + 32
print("T7f: Lahore in °F =", np.round(lahore_f, 1))

monsoon = rain[:, 6:9]                  # Jul, Aug, Sep
monsoon_total = monsoon.sum(axis=1)
print("T7g: Monsoon rain per city (mm) =", monsoon_total)
print("T7g: Wettest monsoon city =", cities[np.argmax(monsoon_total)])

share = monsoon_total / rain.sum(axis=1) * 100
print("T7h: Monsoon share of yearly rain (%) =", np.round(share, 1))
```

```python
# REPLACE: row 0 = FY2024, row 1 = FY2025 (Jul to Jun), values in %
inf = np.array([
    [28.3, 27.4, 31.4, 29.7, 29.7, 28.7, 23.1, 20.7, 17.3, 11.8, 12.6, 12.6],
    [11.1,  9.6,  9.2,  7.2,  4.9,  4.1,  2.4,  1.5,  0.7,  0.3,  3.5,  3.2]])
fy24, fy25 = inf[0], inf[1]

print("T8a: Average inflation (%) [FY24, FY25] =", np.round(inf.mean(axis=1), 2))
print("T8b: FY2024 months above 25 % =", np.sum(fy24 > 25))

change = np.diff(fy24)
print("T8c: Month-to-month change (%) =", np.round(change, 2))
print("T8d: Biggest drop (%) =", change.min(), "| month number =", np.argmin(change) + 2)

label = np.where(fy25 < 5, "Low", np.where(fy25 < 10, "Moderate", "High"))
print("T8e: FY2025 labels =", label)
```

```python
# REPLACE with your class marks
names = np.array(["Ali", "Sara", "Hamza", "Ayesha", "Bilal", "Zainab", "Usman", "Hira"])
marks = np.array([35, 38, 52, 97, 40, 36, 75, 99])

graced = np.clip(marks + 5, 0, 100)
print("T9a: Marks after grace =", graced)

pass_pct = np.sum(graced >= 40) / marks.size * 100
print("T9b: Pass percentage (%) =", pass_pct)

only_grace = (marks < 40) & (graced >= 40)
print("T9c: Passed only due to grace =", names[only_grace])

z = (graced - graced.mean()) / graced.std()
print("T9d: Z-scores =", np.round(z, 2))
```
